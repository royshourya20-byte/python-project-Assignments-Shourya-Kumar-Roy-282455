```python
#Sum and average of first n natural numbers

n = int(input("Enter the number: "))
sum = 0

for i in range(1, n + 1):
    sum += i

average = sum / n

print("Sum =", sum)
print("Average =", average)
```

```python
#Calculating factorial of a number with a loop

n = int(input("Enter a number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
```

```python
#Program that takes a string such as "remember" and reverses it.

s = input("Enter a string: ")
rev = ""

for i in s:
    rev = i + rev

print("Reversed string =", rev)
```

```python
#To check whether a user given input year is leap year or not.

year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```
