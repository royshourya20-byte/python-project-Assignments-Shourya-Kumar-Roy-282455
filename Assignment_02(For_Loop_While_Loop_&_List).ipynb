```python
#For Loop
#Write a Python program to print all the numbers between 1 and 100 that are divisible by both 3 and 5.


for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
```

```python
#For loop
#Write a Python program to take an integer n as input and calculate the sum of the following series:
#1^2 + 2^2 + 3^2 + ... + n^2

n = int(input("Enter a number: "))
sum = 0

for i in range(1, n + 1):
    sum = sum + i ** 2

print("Sum =", sum)
```

```python
#For Loop
#Write a Python program to take a number as input and check whether it is a prime number using a for loop.

n = int(input("Enter a number: "))
count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1

if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
#While Loop
#Write a Python program to take an integer as input and find the sum of its digits using a while loop.

n = int(input("Enter a number: "))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits =", sum)
```

```python
#While Loop
#Write a Python program to take an integer n as input and reverse the number using a while loop.

n = int(input("Enter a number: "))
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print("Reversed number =", rev)
```

```python
#While Loop
#Write a Python program to repeatedly ask the user to enter a number until the user enters 0. At the end, print the sum of all the numbers entered (excluding 0).

sum = 0
n = int(input("Enter a number: "))

while n != 0:
    sum = sum + n
    n = int(input("Enter a number: "))

print("Sum =", sum)
```

```python
#List
#Write a Python program to take 10 numbers from the user and store them in a list. Then find and print the largest and smallest elements without using the max() and min() functions.

a = []

for i in range(10):
    n = int(input("Enter a number: "))
    a.append(n)

largest = a[0]
smallest = a[0]

for i in a:
    if i > largest:
        largest = i
    if i < smallest:
        smallest = i

print("Largest =", largest)
print("Smallest =", smallest)
```

```python
#List
#Write a Python program to remove all duplicate elements from a list without using set().

a = [10, 20, 10, 30, 20, 40]
b = []

for i in a:
    if i not in b:
        b.append(i)

print("Original list =", a)
print("List without duplicates =", b)
```

```python
#List
#Write a Python program to take a list of numbers and create two separate lists:
#- One containing all even numbers
#- One containing all odd numbers

a = [12, 7, 5, 18, 20, 9]
even = []
odd = []

for i in a:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even:", even)
print("Odd:", odd)
```
