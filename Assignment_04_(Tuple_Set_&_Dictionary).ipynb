```python
#Tuple
#Remove Duplicate Elements:
#Given a tuple containing duplicate values, create a new tuple containing only unique elements while preserving their original order.
#Example: (1, 2, 2, 3, 1, 4, 3) → (1, 2, 3, 4)

t = (1, 2, 2, 3, 1, 4, 3)
new = ()

for i in t:
    if i not in new:
        new = new + (i,)

print("Original tuple:", t)
print("Unique tuple:", new)
```

```python
#Tuple
#2. Sort a Tuple of Tuples:
#Given students = (("Riya", 85), ("Aman", 92), ("Soham", 78), ("Priya", 89)), write a program to sort the students in descending order of marks and display their names and marks.

students = (("Riya", 85), ("Aman", 92), ("Soham", 78), ("Priya", 89))

sorted_students = sorted(students, key=lambda x: x[1], reverse=True)

for name, marks in sorted_students:
    print(name, marks)
```

```python
#Tuple
#Write a Python program to create a tuple of 5 integers entered by the user. Calculate and display the sum and average of its elements.

t = ()

for i in range(5):
    n = int(input("Enter a number: "))
    t = t + (n,)

total = sum(t)
average = total / 5

print("Tuple:", t)
print("Sum =", total)
print("Average =", average)
```

```python
#Tuple
#Count Occurrences:
#Given a tuple t = (10, 20, 10, 30, 10, 40, 20), write a program to count how many times the number 10 occurs in the tuple.

t = (10, 20, 10, 30, 10, 40, 20)

count = t.count(10)

print("Number of occurrences of 10 =", count)
```

```python
#Tuple
#Find Maximum and Minimum:
#Write a program to find the largest and smallest elements in a tuple without using the built-in max() and min() functions.

t = (10, 25, 5, 40, 15, 2)

largest = t[0]
smallest = t[0]

for i in t:
    if i > largest:
        largest = i
    if i < smallest:
        smallest = i

print("Largest =", largest)
print("Smallest =", smallest)
```

```python
#Set
#Given two sets, A = {1, 2, 3, 4, 5} and B = {4, 5, 6, 7, 8}, display their union, intersection, and difference (A - B).

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
```

```python
#Set
#Common Elements in Two Lists
#Write a program to accept two lists of integers and display all distinct elements that appear in both lists using sets.

a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]

A = set(a)
B = set(b)

common = A & B

print("Common elements:", common)

```

```python
#Set
#Check Subset and Superset
#Given two sets, write a program to check whether the first set is a subset of the second set and whether the second set is a superset of the first.

A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
#Dictionary
#Student Marks Dictionary
#Create a dictionary containing the names and marks of 5 students. Display the name and marks of each student using a loop.

students = {
    "Riya": 85,
    "Aman": 92,
    "Soham": 78,
    "Priya": 89,
    "Rahul": 80
}

for name, marks in students.items():
    print(name, ":", marks)
```

```python
#Dictionary
#Frequency of Words
#Write a Python program to accept a sentence and create a dictionary containing the frequency of each word.
#Example: "python is easy and python is useful"
#Expected result: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'useful': 1}

sentence = input("Enter a sentence: ")

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
#Dictionary
#Merge Two Dictionaries and Add Common Values
#Given two dictionaries containing item names and quantities, merge them into a single dictionary. If an item appears in both dictionaries, add its quantities.
#Example:
#d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
#d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}
#Expected result: {"Pen": 14, "Pencil": 12, "Book": 3, "Eraser": 6}

d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print("Merged dictionary:", result)
```
